```python
import os
import random
import numpy as np
from PIL import Image, ImageDraw, ImageFont

# 생성할 이미지 수
NUM_IMAGES = 10
# 이미지 크기
IMG_SIZE = (500, 500)

# 폰트 경로 (시스템 폰트 또는 손글씨 TTF 파일 경로 지정)
# 윈도우 기본 폰트 예시: arial.ttf 또는 malgun.ttf
FONT_PATH = "arial.ttf"

output_dir = "./generated_images"
os.makedirs(output_dir, exist_ok=True)

for img_idx in range(NUM_IMAGES):
    # 흰색 배경 이미지 생성
    img = Image.new("RGB", IMG_SIZE, color="white")
    draw = ImageDraw.Draw(img)

    # 한 이미지당 무작위 개수(10~15개)의 숫자 배치
    num_digits = random.randint(10, 15)

    for _ in range(num_digits):
        digit = str(random.randint(0, 9))

        # 랜덤 글자 크기, 위치, 색상 지정
        font_size = random.randint(20, 70)
        x = random.randint(20, IMG_SIZE[0] - 60)
        y = random.randint(20, IMG_SIZE[1] - 60)

        # 검은색 또는 빨간색 랜덤 선택 (빨간색 비중 약 30%)
        is_red = random.random() < 0.3
        color = (220, 20, 60) if is_red else (20, 20, 20)

        try:
            font = ImageFont.truetype(FONT_PATH, font_size)
        except OSError:
            font = ImageFont.load_default()

        # 숫자 그리기
        draw.text((x, y), digit, fill=color, font=font)

    # 이미지 파일 저장
    save_path = os.path.join(output_dir, f"random_digits_{img_idx+1:02d}.jpg")
    img.save(save_path)
    print(f"저장 완료: {save_path}")
```
